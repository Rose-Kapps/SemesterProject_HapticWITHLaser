```python
import csv
import time
import plotly.graph_objects as go
from IPython.display import display, clear_output

# Function to read new data from the CSV file
def read_new_data():
    new_x = []
    new_y = []
    
    try:
        with open("output.csv", "r") as file:
            reader = csv.reader(file)
            next(reader)  # Skip header

            for line in reader:
                if len(line) == 2:  # Ensure proper format (two columns)
                    try:
                        new_x.append(float(line[0]))
                        new_y.append(float(line[1]))
                    except ValueError as e:
                        print(f"Error parsing line {line}: {e}")
                else:
                    print(f"Skipping malformed line: {line}")

    except IOError as e:
        print(f"Error reading file: {e}")
    except Exception as e:
        print(f"Unexpected error: {e}")

    return new_x, new_y

# Create a figure object
fig = go.Figure()

# Function to update the plot
def update_graph():
    new_x, new_y = read_new_data()

    # Only update the graph if there is new data
    if new_x and new_y:
        # Sort the data based on the x values
        sorted_data = sorted(zip(new_x, new_y))  # Zip the data, sort by x-coordinate

        # Unzip the sorted data back into x and y lists
        new_x, new_y = zip(*sorted_data)  # Unzip sorted pairs back into separate lists
        
        # Convert tuples back to lists
        new_x = list(new_x)
        new_y = list(new_y)

        print(f"Sorted data: {new_x}, {new_y}")
        
        # Clear the previous plot (if any)
        clear_output(wait=True)

        # Update the plot
        fig.data = []  # Clear existing traces
        fig.add_trace(go.Scatter(x=new_x, y=new_y, mode='lines+markers', name='Data'))

        # Customize layout (optional)
        fig.update_layout(
            title="Real-Time Data Plot",
            xaxis_title="X Axis",
            yaxis_title="Y Axis",
            showlegend=True
        )

        # Display the updated figure
        display(fig)

    else:
        print("No new data to update")

# Periodically check for new data every second
while True:
    update_graph()
    time.sleep(0.1)  # Adjust the delay as needed
```
